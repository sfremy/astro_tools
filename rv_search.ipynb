```python
import pandas as pd
import os
```

```python
names = [17051, 7788, 33564, 182572, 160032, 131156, 25998, 156897, 20794, 19373, 219623, 166620]

for name in names:
    print('Searching for: HD' + str(name))
    for file in os.listdir('harps_rv_bank'):
        if file.startswith('HD ' + str(name)):
            print('HARPS RV Bank match: ' + file)

    for file in os.listdir('keck_vels'):
        if file.startswith('HD ' + str(name)):
            print('Keck HIRES match: ' + file)

    for file in os.listdir('CLS_Rosenthal_2021'):
        if file.startswith('HD ' + str(name)):
            print('CLS match: ' + file)

    for file in os.listdir('Lick_Hamilton_F14'):
        if file.startswith('HD ' + str(name)):
            print('Lick/Hamilton match:' + file)

    for file in os.listdir('Laliotis_2023'):
        if file.startswith('HD ' + str(name)):
            print('Laliotis et al. 2023 match:' + file)

    print('-----------------------------------')
    
    #Note - Don't take Keck & HIRES data from CLS if you can help it
```
