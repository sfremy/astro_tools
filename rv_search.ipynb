```python
from astroquery.simbad import Simbad
import time
import os

folder = 'Lick_Hamilton_F14'

target = 'GL'

for file in os.listdir(folder):
    if file.startswith(target):
        try:
            obj_id = file.split('.')[0].replace('L','J')
            j_k = file.split('.')[1].split('.')[0]
            full_path = os.path.join(folder, file)

            result_table = Simbad.query_objectids(obj_id)
            id_list = result_table['ID'].tolist()
            matching_id = next((obj_id for obj_id in id_list if 'HD' in obj_id), None)

            if matching_id:
                print('Found HD catalog entry')
                new_name = matching_id + '_' + j_k
                os.rename(full_path, os.path.join(folder, new_name))
            else:
                os.remove(full_path)

            time.sleep(0.1)

        except:
            print('Error with file: ' + file)
            continue


#Unclear - g & s stars, HATS catalog stars
```
