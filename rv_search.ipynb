```python
import pandas as pd
import numpy as np

#NOTE - Lick/Hamilton received unspecified upgrades between 1992 and 1995.

STD_COL_NAMES = ['HD', 'JD', 'RVel', 'e_RVel', 'Sind', 'e_Sind', 'Hind', 'e_Hind', 'EWHa', 'e_EWHa', 'RHKp', 'e_RHKp', 'BIS', 'FWHM', 'Tel', 'Bibcode']

def get_EBPS_HIRES(data_file, SYSTEM_HD):
    """Read in a .vels data file from the EBPS HIRES RV archive

    Args:
        data_file (str): path to .vels file
        SYSTEM_HD (int): HD number of the system

    Returns:
        pd.DataFrame: Pandas dataframe with the standard column names 
        containing HIRES/Keck RV data
    """

    # load .vels file
    col_names = ['JD', 'RVel', 'e_RVel', 'S_value', 'Halpha', \
                'phot_per_pix', 't_exp']
    raw = pd.read_csv('ebps_keck_hires/' + data_file, sep='\s+', header=None, names=col_names)

    # create new pandas df
    new = pd.DataFrame()

    new = pd.concat([new, raw[['JD', 'RVel', 'e_RVel', 'S_value', 'Halpha']].astype(float)])
    
    # add missing standard column names
    new = new.rename(columns={'S_value':'Sind', 'Halpha':'Hind'})

    new['HD'] = SYSTEM_HD
    new['Bibcode'] = '2017yCat..51530208B'
    #JD
    #RVel
    #e_RVel
    new['Sind'] = new['Sind'].fillna(-1)
    new['e_Sind'] = -1
    new['Hind'] = new['Hind'].fillna(-1)
    new['e_Hind'] = -1
    new['EWHa'] = -1
    new['e_EWHa'] = -1
    new['RHKp'] = -1
    new['e_RHKp'] = -1
    new['BIS'] = 0
    new['FWHM'] = -1
    new['Tel'] = np.where(new['JD'] < 2453236, 'HIRES-pre', 'HIRES-post')
    
    # reorder columns to standard format
    new = new[STD_COL_NAMES]

    new.dropna()
    return new

def get_HARPS_RVBank(data_file, SYSTEM_HD):
    """Read in a .csv data file from the HARPS RVBank archive

    Args:
        data_file (str): path to .csv file
        SYSTEM_HD (int): HD number of the system

    Returns:
        pd.DataFrame: Pandas dataframe with the standard column names 
        containing HARPS RVBank data
    """

    raw = pd.read_csv('harps_rv_bank/' + data_file)
    new = pd.DataFrame()
    
    new = pd.concat([new,raw[['BJD', 'RV_mlc_nzp', 'e_RV_mlc_nzp', 'Halpha', 'e_Halpha', 'RHKp', 'dRHKp', 'BIS', 'FWHM_DRS']].astype(float)])
    new = new.rename(columns = {'BJD':'JD', 'RV_mlc_nzp':'RVel', 'e_RV_mlc_nzp':'e_RVel', 'Halpha':'Hind', 
                                'e_Halpha':'e_Hind', 'dRHKp':'e_RHKp', 'FWHM_DRS':'FWHM'})

    new['HD'] = SYSTEM_HD
    new['Bibcode'] = '2020yCat..36360074T'
    #JD
    #RVel
    #e_RVel
    new['Sind'] = -1
    new['e_Sind'] = -1
    new['Hind'] = new['Hind'].fillna(-1)
    new['e_Hind'] = new['e_Hind'].fillna(-1)
    new['EWHa'] = -1
    new['e_EWHa'] = -1
    new['RHKp'] = new['RHKp'].fillna(-1)
    new['e_RHKp'] = new['RHKp'].fillna(-1)
    new['BIS'] = (new['BIS'] * 1000).fillna(0)
    new['FWHM'] = (new['FWHM'] * 1000).fillna(-1)
    new['Tel'] = np.where(new['JD'] < 2457163, 'HARPS-pre', 'HARPS-post')

    new = new[STD_COL_NAMES]

    new.dropna()
    return new

def get_CLS(data_file, SYSTEM_HD):
    """Read in a .csv data file from the CLS archive

    Args:
        data_file (str): path to .csv file
        SYSTEM_HD (int): HD number of the system

    Returns:
        pd.DataFrame: Pandas dataframe with the standard column names 
        containing CLS data
    """

    raw = pd.read_csv('CLS_Rosenthal_2021/' + data_file)
    new = pd.DataFrame()

    new = pd.concat([new, raw[['BJD', 'RVel', 'e_RVel', 'SVal']].astype(float)])
    new = new.rename(columns = {'BJD':'JD', 'SVal':'Sind'})

    new['HD'] = SYSTEM_HD
    new['Bibcode'] = '2021ApJS..255....8R'
    #JD
    #RVel
    #e_RVel
    new['Sind'] = new['Sind'].fillna(-1)
    new['e_Sind'] = -1
    new['Hind'] = -1
    new['e_Hind'] = -1
    new['EWHa'] = -1
    new['e_EWHa'] = -1
    new['RHKp'] = -1
    new['e_RHKp'] = -1
    new['BIS'] = 0
    new['FWHM'] = -1

    if 'Levy' in data_file:
        new['Tel'] = 'Levy'
    if 'Hamilton' in data_file:
        new['Tel'] = 'Hamilton'
    if 'HIRES' in data_file:
        new['Tel'] = np.where(new['JD'] < 2453236, 'HIRES-pre', 'HIRES-post')

    new.dropna()
    new = new[STD_COL_NAMES]

    return new

def get_Lick_Hamilton(data_file, SYSTEM_HD):
    """Read in a .csv data file from the Lick/Hamilton archive

    Args:
        data_file (str): path to .csv file
        SYSTEM_HD (int): HD number of the system

    Returns:
        pd.DataFrame: Pandas dataframe with the standard column names 
        containing Lick/Hamilton data
    """

    raw = pd.read_csv('Lick_Hamilton_F14/' + data_file)
    new = pd.DataFrame()

    new = pd.concat([new, raw[['JD', 'RVel', 'e_RVel']].astype(float)])

    new['HD'] = SYSTEM_HD
    new['Bibcode'] = '2014yCat..22100005F'
    #JD
    #RVel
    #e_RVel
    new['Sind'] = -1
    new['e_Sind'] = -1
    new['Hind'] = -1
    new['e_Hind'] = -1
    new['EWHa'] = -1
    new['e_EWHa'] = -1
    new['RHKp'] = -1
    new['e_RHKp'] = -1
    new['BIS'] = 0
    new['FWHM'] = -1
    new['Tel'] = 'Hamilton'

    new.dropna()
    new = new[STD_COL_NAMES]

    return new

def get_Laliotis(data_file, SYSTEM_HD):
    """Read in a .csv data file from the Laliotis archive

    Args:
        data_file (str): path to .csv file
        SYSTEM_HD (int): HD number of the system

    Returns:
        pd.DataFrame: Pandas dataframe with the standard column names 
        containing Laliotis data
    """
        
    raw = pd.read_csv('Laliotis_2023/' + data_file)
    new = pd.DataFrame()

    new = pd.concat([new, raw[['BJD', 'RVel', 'e_RVel', 'SHK', 'e_SHK', 'EWHa', 'e_EWHa']].astype(float), raw['r_RVel']])
    new = new.rename(columns={'BJD':'JD', 'r_RVel':'Tel', 'SHK':'Sind', 'e_SHK':'e_Sind'})

    new['HD'] = SYSTEM_HD
    new['Bibcode'] = '2023yCat..51650176L'
    #JD
    #RVel
    #e_RVel
    new['Sind'] = new['Sind'].fillna(-1)
    new['e_Sind'] = new['e_Sind'].fillna(-1)
    new['Hind'] = -1
    new['e_Hind'] = -1
    new['EWHa'] = new['EWHa'].fillna(-1)
    new['e_EWHa'] = new['e_EWHa'].fillna(-1)
    new['RHKp'] = -1
    new['e_RHKp'] = -1
    new['BIS'] = 0
    new['FWHM'] = -1

    if 'PFS-Pre' in data_file:
        new['Tel'] = 'PFS-Pre'
    if 'PFS-Post' in data_file:
        new['Tel'] = 'PFS-Post'
    if 'UCLES' in data_file:
        new['Tel'] = 'UCLES'
        

    new.dropna()
    new = new[STD_COL_NAMES]

    return new

def get_DACE(filename, SYSTEM_HD):
    raw = pd.read_csv('dace_timeseries/' + filename, sep='\t', comment='#')
    new = pd.DataFrame()

    if 'CES' in filename:
        new['JD'] = raw['rjd'][1:].astype(float) + 2400000
        new['RVel'] = raw['vrad'][1:].astype(float)
        new['e_RVel'] = raw['svrad'][1:].astype(float)
        new['FWHM'] = -1
        new['Tel'] = np.where(raw['rjd'][1:].astype(float) + 2400000 < 2450912.5, 'CES-pre', 'CES-post')

        new['Bibcode'] = '2000A&A...353L..33K'

    if 'CORALIE' in filename:
        new['JD'] = raw['rjd'][1:].astype(float) + 2400000
        new['RVel'] = raw['vrad'][1:].astype(float)
        new['e_RVel'] = raw['svrad'][1:].astype(float)
        new['FWHM'] = -1
        new['Tel'] = np.where(raw['rjd'][1:].astype(float) + 2400000 < 2454256.5, 'CORALIE-98', 'CORALIE-07')

        new['Bibcode'] = '2001A&A...375..205N'

    if '2001_UCLES' in filename:
        new['JD'] = raw['rjd'][1:].astype(float) + 2400000
        new['RVel'] = raw['vrad'][1:].astype(float)
        new['e_RVel'] = raw['svrad'][1:].astype(float)
        new['FWHM'] = -1
        new['Tel'] = 'UCLES'

        new['Bibcode'] = '2001ApJ...555..410B'

    if '2006_UCLES' in filename:
        new['JD'] = raw['rjd'][1:].astype(float) + 2400000
        new['RVel'] = raw['vrad'][1:].astype(float)
        new['e_RVel'] = raw['svrad'][1:].astype(float)
        new['FWHM'] = -1
        new['Tel'] = 'UCLES'

        new['Bibcode'] = '2006ApJ...646..505B'

    if 'ELODIE' in filename:
        new['JD'] = raw['rjd'][1:].astype(float) + 2400000
        new['RVel'] = raw['vrad'][1:].astype(float)
        new['e_RVel'] = raw['svrad'][1:].astype(float)
        new['FWHM'] = raw['fwhm'][1:].astype(float).fillna(-1)
        new['Tel'] = 'ELODIE'

        new['Bibcode'] = '2004PASP..116..693M'

    new['Sind'] = -1
    new['e_Sind'] = -1
    new['Hind'] = -1
    new['e_Hind'] = -1
    new['EWHa'] = -1
    new['e_EWHa'] = -1
    new['RHKp'] = -1
    new['e_RHKp'] = -1
    new['BIS'] = 0
    new['HD'] = SYSTEM_HD

    new.dropna()
    new = new[STD_COL_NAMES]

    return new

def get_SOPHIE(filename, SYSTEM_HD):
    raw = pd.read_csv(
        'SOPHIE_catalog/' + filename,
        sep=r'\s+',
        comment='#',
        header=None,
        engine='python'
    )

    # Set column names based on the header info in the file
    raw.columns = [
        'seq', 'objname', 'date', 'slen', 'mask', 'ccf_offline', 'rv', 'err',
        'fwhm', 'span', 'maxcpp', 'contrast', 'lines',
        'view_head', 'get_ccf', 'view_ccf', 'bjd'
    ]

    new = pd.DataFrame()

    new['JD'] = raw['bjd'][1:].astype(float)
    new['RVel'] = raw['rv'][1:].astype(float) * 1000
    new['e_RVel'] = raw['err'][1:].astype(float) * 1000
    new['FWHM'] = raw['fwhm'][1:].astype(float) * 1000
    new['Tel'] = np.where(raw['bjd'][1:].astype(float) < 2457700, 'SOPHIE-pre', 'SOPHIE-post')
    new['Sind'] = -1
    new['e_Sind'] = -1
    new['Hind'] = -1
    new['e_Hind'] = -1
    new['EWHa'] = -1
    new['e_EWHa'] = -1
    new['RHKp'] = -1
    new['e_RHKp'] = -1
    new['BIS'] = 0
    new['HD'] = SYSTEM_HD
    new['Bibcode'] = '2004PASP..116..693M'

    new.dropna()
    new = new[STD_COL_NAMES]

    return new
```

```python
import os
import shutil

names = [74576, 78366, 78154, 88230, 90589]

for name in names:
    harps_rvs = pd.DataFrame()
    hires_rvs = pd.DataFrame()
    hamilton_rvs = pd.DataFrame()
    sophie_rvs = pd.DataFrame()

    print('Searching for: HD' + str(name))
    for file in os.listdir('harps_rv_bank'):
        if str(name) + '_' in file:
            print('HARPS RV Bank match: ' + file)
            harps_rvs = get_HARPS_RVBank(file, name)
            shutil.copy('harps_rv_bank/' + file, 'output_csvs')

    for file in os.listdir('ebps_keck_hires'):
        if str(name) + '_' in file:
            print('Keck HIRES match: ' + file)
            hires_rvs = get_EBPS_HIRES(file, name)
            shutil.copy('ebps_keck_hires/' + file, 'output_csvs')

    for file in os.listdir('Lick_Hamilton_F14'):
        if str(name) + '_' in file:
            print('Lick/Hamilton match: ' + file)
            hamilton_rvs = get_Lick_Hamilton(file, name)
            hamilton_filename = file

    rosenthal_files = pd.DataFrame()
    for file in os.listdir('CLS_Rosenthal_2021'):
        if str(name) + '_' in file:
            cls_hamilton = pd.DataFrame()
            print('CLS match: ' + file)
            if hires_rvs.empty and 'HIRES' in file:
                    rosenthal_files = pd.concat([rosenthal_files, get_CLS(file, name)])
                    shutil.copy('CLS_Rosenthal_2021/' + file, 'output_csvs')
            if 'Hamilton' in file:
                cls_hamilton = get_CLS(file, name)
            if 'Levy' in file:
                rosenthal_files = pd.concat([rosenthal_files, get_CLS(file, name)])
                shutil.copy('CLS_Rosenthal_2021/' + file, 'output_csvs')
            if not hamilton_rvs.empty:
                if not cls_hamilton.empty:
                    if len(cls_hamilton.index) > len(hamilton_rvs.index):
                        hamilton_rvs = pd.DataFrame()
                        rosenthal_files = pd.concat([rosenthal_files, cls_hamilton])
                        shutil.copy('CLS_Rosenthal_2021/' + file, 'output_csvs')
                    else:
                        shutil.copy('Lick_Hamilton_F14/' + hamilton_filename, 'output_csvs')
                else:
                    shutil.copy('Lick_Hamilton_F14/' + hamilton_filename, 'output_csvs')
            elif not cls_hamilton.empty:
                shutil.copy('CLS_Rosenthal_2021/' + file, 'output_csvs')

    laliotis_files = pd.DataFrame()
    for file in os.listdir('Laliotis_2023'):
        if str(name) + '_' in file:
            print('Laliotis et al. 2023 match: ' + file)
            laliotis_files = pd.concat([laliotis_files, get_Laliotis(file, name)])
            shutil.copy('Laliotis_2023/' + file, 'output_csvs')

    dace_files = pd.DataFrame()
    for file in os.listdir('dace_timeseries'):
        if str(name) + '_' in file:
            print('DACE match: ' + file)
            dace_files = pd.concat([dace_files, get_DACE(file, name)])
            shutil.copy('dace_timeseries/' + file, 'output_csvs')
        
    for file in os.listdir('SOPHIE_catalog'):
        if str(name) + '_' in file:
            print('SOPHIE match: ' + file)
            sophie_rvs = get_SOPHIE(file, name)
            shutil.copy('SOPHIE_catalog/' + file, 'output_csvs')

    dataframes = [harps_rvs, hires_rvs, hamilton_rvs, rosenthal_files, laliotis_files, dace_files, sophie_rvs]
    valid_dataframes = [df for df in dataframes if df is not None]

    if valid_dataframes:
        combined_df = pd.concat(valid_dataframes)
        if combined_df.empty:
            print('Nothing found.')
        else:
            combined_df.to_csv(f'output_csvs/{name}_main_data_all.csv')
            print(np.unique(np.array(combined_df['Bibcode'])))
    else:
        print("Warning: No valid dataframes to concatenate.")
    print('-----------------------------------')
```

```python
def get_nari(filename):
    if 'DRS' in filename:
        print('DRS')
        raw = pd.read_csv(
            'Nari_2025/' + filename,
            sep=r'\s+',
            comment='#',
            header=None,
            engine='python',
            skiprows = 54
        )

        raw.columns = ['JD', 'RVel', 'e_RVel', 'FWHM', 'e_FWHM', 'BisSpan', 'e_BisSpan', 'Contrast', 'e_Contrast', 'Sind', 'e_Sind', 'Hind', 'e_Hind', 'Na', 'e_Na', 'BERV']
        raw = raw[3:-1]

        new = pd.DataFrame()

        new['JD'] = raw['JD'].astype(float) + 2500000
        new['RVel'] = raw['RVel'].astype(float)
        new['e_RVel'] = raw['e_RVel'].astype(float)

        new['Sind'] = raw['Sind'].astype(float)
        new['e_Sind'] = raw['e_Sind'].astype(float)
        new['Hind'] = raw['Hind'].astype(float)
        new['e_Hind'] = raw['e_Hind'].astype(float)
        new['FWHM'] = raw['FWHM'].astype(float)

        new['Tel'] = np.where(raw['JD'].astype(float) < 2458673.5, 'ESPRESSO-Pre', 'ESPRESSO-Post')
        new['Bibcode'] = '2025A&A...693A.297N'
        new['HD'] = 20794

        new['EWHa'] = -1
        new['e_EWHa'] = -1
        new['RHKp'] = -1
        new['e_RHKp'] = -1
        new['BIS'] = 0

        new = new[STD_COL_NAMES]

        return new

    if 'SBART' in filename:    
        print('SBART')
        raw = pd.read_csv(
            'Nari_2025/' + filename,
            sep=r'\s+',
            comment='#',
            header=None,
            engine='python',
            skiprows = 54
        )

        raw.columns = ['JD', 'RVel', 'e_RVel']
        raw = raw[:-1]

        raw['e_RVel'] = raw['e_RVel'].str.replace('\\', '', regex = False)

        new = pd.DataFrame()

        new['JD'] = raw['JD'].astype(float)
        new['RVel'] = raw['RVel'].astype(float)
        new['e_RVel'] = raw['e_RVel'].astype(float)

        new['Sind'] = -1
        new['e_Sind'] = -1
        new['Hind'] = -1
        new['e_Hind'] = -1
        new['FWHM'] = -1
        new['EWHa'] = -1
        new['e_EWHa'] = -1
        new['RHKp'] = -1
        new['e_RHKp'] = -1
        new['BIS'] = 0

        new['Tel'] = np.where(raw['JD'].astype(float) < 2458673.5, 'ESPRESSO-Pre', 'ESPRESSO-Post')
        new['Bibcode'] = '2025A&A...693A.297N'
        new['HD'] = 20794

        new = new[STD_COL_NAMES]    

        return new
        
    if 'HARPS-1' in filename:
        print('HARPS')
        raw = pd.read_csv(
            'Nari_2025/' + filename,
            sep=r'\s+',
            comment='#',
            header=None,
            engine='python',
            skiprows = 54
        )

        raw.columns = ['JD', 'RVel', 'e_RVel', 'FWHM', 'e_FWHM', 'BisSpan', 'e_BisSpan', 'Contrast', 'e_Contrast', 'Sind', 'e_Sind', 'Hind', 'e_Hind', 'NaI', 'e_NaI', 'eTemp', 'e_eTemp']
        raw = raw[3:-1].dropna()

        new = pd.DataFrame()

        new['JD'] = raw['JD'].astype(float)
        new['RVel'] = raw['RVel'].astype(float)
        new['e_RVel'] = raw['e_RVel'].astype(float)

        new['Sind'] = raw['Sind'].astype(float)
        new['e_Sind'] = raw['e_Sind'].astype(float)
        new['Hind'] = raw['Hind'].astype(float)
        new['e_Hind'] = raw['e_Hind'].astype(float)
        new['FWHM'] = raw['FWHM'].astype(float)

        new['EWHa'] = -1
        new['e_EWHa'] = -1
        new['RHKp'] = -1
        new['e_RHKp'] = -1
        new['BIS'] = 0

        new['Tel'] = np.where(raw['JD'].astype(float) < 2457163, 'HARPS-pre', 'HARPS-post')
        new['Bibcode'] = '2025A&A...693A.297N'
        new['HD'] = 20794    

        new = new[STD_COL_NAMES]

        return new
    
    if 'HARPS-2' in filename:
        print('HARPS')
        raw = pd.read_csv(
            'Nari_2025/' + filename,
            sep=r'\s+',
            comment='#',
            header=None,
            engine='python',
            skiprows = 54
        )

        raw.columns = ['JD', 'RVel', 'e_RVel', 'FWHM', 'e_FWHM', 'BisSpan', 'e_BisSpan', 'Contrast', 'e_Contrast', 'Sind', 'e_Sind', 'Hind', 'e_Hind', 'NaI', 'e_NaI', 'BERV']
        raw = raw[3:-1].dropna()

        new = pd.DataFrame()

        new['JD'] = raw['JD'].astype(float)
        new['RVel'] = raw['RVel'].astype(float)
        new['e_RVel'] = raw['e_RVel'].astype(float)

        new['Sind'] = raw['Sind'].astype(float)
        new['e_Sind'] = raw['e_Sind'].astype(float)
        new['Hind'] = raw['Hind'].astype(float)
        new['e_Hind'] = raw['e_Hind'].astype(float)
        new['FWHM'] = raw['FWHM'].astype(float)

        new['EWHa'] = -1
        new['e_EWHa'] = -1
        new['RHKp'] = -1
        new['e_RHKp'] = -1
        new['BIS'] = 0

        new['Tel'] = np.where(raw['JD'].astype(float) < 2457163, 'HARPS-pre', 'HARPS-post')
        new['Bibcode'] = '2025A&A...693A.297N'
        new['HD'] = 20794    

        new = new[STD_COL_NAMES]

        return new
    
    if 'HARPS-3' in filename:
        print('HARPS')
        raw = pd.read_csv(
            'Nari_2025/' + filename,
            sep=r'\s+',
            comment='#',
            header=None,
            engine='python',
            skiprows = 54
        )

        raw.columns = ['JD', 'RVel', 'e_RVel', 'FWHM', 'e_FWHM', 'BisSpan', 'e_BisSpan', 'Contrast', 'e_Contrast', 'Sind', 'e_Sind', 'Hind', 'e_Hind', 'NaI', 'e_NaI', 'BERV']
        raw = raw[3:-1].dropna()

        new = pd.DataFrame()

        new['JD'] = raw['JD'].astype(float)
        new['RVel'] = raw['RVel'].astype(float)
        new['e_RVel'] = raw['e_RVel'].astype(float)

        new['Sind'] = raw['Sind'].astype(float)
        new['e_Sind'] = raw['e_Sind'].astype(float)
        new['Hind'] = raw['Hind'].astype(float)
        new['e_Hind'] = raw['e_Hind'].astype(float)
        new['FWHM'] = raw['FWHM'].astype(float)

        new['EWHa'] = -1
        new['e_EWHa'] = -1
        new['RHKp'] = -1
        new['e_RHKp'] = -1
        new['BIS'] = 0

        new['Tel'] = np.where(raw['JD'].astype(float) < 2457163, 'HARPS-pre', 'HARPS-post')
        new['Bibcode'] = '2025A&A...693A.297N'
        new['HD'] = 20794    

        new = new[STD_COL_NAMES]

        return new
```
