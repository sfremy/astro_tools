```python
import pandas as pd
import numpy as np

#NOTE - Lick/Hamilton received unspecified upgrades between 1992 and 1995.

STD_COL_NAMES = ['HD', 'JD', 'RVel', 'e_RVel', 
                 'Sind', 'e_Sind', 'Hind', 'e_Hind', 
                 'EWHa', 'e_EWHa', 'RHKp', 'e_RHKp', 
                 'BIS', 'FWHM', 'Tel', 'Bibcode']

def get_EBPS_HIRES(data_file, SYSTEM_HD):
    """Read in a .vels data file from the EBPS HIRES RV archive

    Args:
        data_file (str): path to .vels file
        SYSTEM_HD (int): HD number of the system

    Returns:
        pd.DataFrame: Pandas dataframe with the standard column names 
        containing HIRES/Keck RV data
    """

    # load .vels file
    col_names = ['JD', 'RVel', 'e_RVel', 'S_value', 'Halpha', \
                'phot_per_pix', 't_exp']
    raw = pd.read_csv(data_file, sep='\s+', header=None, names=col_names)

    # create new pandas df
    new = pd.DataFrame()

    new = pd.concat([new, raw[['JD', 'RVel', 'e_RVel', 'S_value', 'Halpha']]])
    
    # add missing standard column names
    new = new.rename(columns={'S_value':'Sind', 'Halpha':'Hind'})

    new['HD'] = SYSTEM_HD
    new['Bibcode'] = '2017yCat..51530208B'
    #JD
    #RVel
    #e_RVel
    new['Sind'] = new['Sind'].fillna(-1)
    new['e_Sind'] = -1
    new['Hind'] = new['Hind'].fillna(-1)
    new['e_Hind'] = -1
    new['EWHa'] = -1
    new['e_EWHa'] = -1
    new['RHKp'] = -1
    new['e_RHKp'] = -1
    new['BIS'] = 0
    new['FWHM'] = -1
    new['Tel'] = np.where(new['JD'] < 2453236, 'HIRES-pre', 'HIRES-post')
    
    # reorder columns to standard format
    new = new[STD_COL_NAMES]

    return new

def get_HARPS_RVBank(data_file, SYSTEM_HD):
    """Read in a .csv data file from the HARPS RVBank archive

    Args:
        data_file (str): path to .csv file
        SYSTEM_HD (int): HD number of the system

    Returns:
        pd.DataFrame: Pandas dataframe with the standard column names 
        containing HARPS RVBank data
    """

    raw = pd.read_csv('harps_rv_bank/' + data_file)
    new = pd.DataFrame()
    
    new = pd.concat([new,raw[['BJD', 'RV_mlc_nzp', 'e_RV_mlc_nzp', 'Halpha', 'e_Halpha', 'RHKp', 'dRHKp', 'BIS', 'FWHM_DRS']]])
    new = new.rename(columns = {'BJD':'JD', 'RV_mlc_nzp':'RVel', 'e_RV_mlc_nzp':'e_RVel', 'Halpha':'Hind', 
                                'e_Halpha':'e_Hind', 'dRHKp':'e_RHKp', 'FWHM_DRS':'FWHM'})

    new['HD'] = SYSTEM_HD
    new['Bibcode'] = '2020yCat..36360074T'
    #JD
    #RVel
    #e_RVel
    new['Sind'] = -1
    new['e_Sind'] = -1
    new['Hind'] = new['Hind'].fillna(-1)
    new['e_Hind'] = new['e_Hind'].fillna(-1)
    new['EWHa'] = -1
    new['e_EWHa'] = -1
    new['RHKp'] = new['RHKp'].fillna(-1)
    new['e_RHKp'] = new['RHKp'].fillna(-1)
    new['BIS'] = (new['BIS'] * 1000).fillna(0)
    new['FWHM'] = (new['FWHM'] * 1000).fillna(-1)
    new['Tel'] = np.where(new['JD'] < 2457163, 'HARPS-pre', 'HARPS-post')

    new = new[STD_COL_NAMES]

    return new

def get_CLS(data_file, SYSTEM_HD):
    """Read in a .csv data file from the CLS archive

    Args:
        data_file (str): path to .csv file
        SYSTEM_HD (int): HD number of the system

    Returns:
        pd.DataFrame: Pandas dataframe with the standard column names 
        containing CLS data
    """

    raw = pd.read_csv('CLS_Rosenthal_2021/' + data_file)
    new = pd.DataFrame()

    new = pd.concat([new, raw[['BJD', 'RVel', 'e_RVel', 'SVal']]])
    new = new.rename(columns = {'BJD':'JD', 'SVal':'Sind'})

    new['HD'] = SYSTEM_HD
    new['Bibcode'] = '2021ApJS..255....8R'
    #JD
    #RVel
    #e_RVel
    new['Sind'] = new['Sind'].fillna(-1)
    new['e_Sind'] = -1
    new['Hind'] = -1
    new['e_Hind'] = -1
    new['EWHa'] = -1
    new['e_EWHa'] = -1
    new['RHKp'] = -1
    new['e_RHKp'] = -1
    new['BIS'] = 0
    new['FWHM'] = -1

    if 'apf' in data_file:
        new['Tel'] = 'Levy'
    elif 'lick' in data_file:
        new['Tel'] = 'Hamilton'
    elif 'j' or 'k' in data_file:
        new['Tel'] = np.where(new['JD'] < 2453236, 'HIRES-pre', 'HIRES-post')

    new = new[STD_COL_NAMES]

    return new

def get_Lick_Hamilton(data_file, SYSTEM_HD):
    """Read in a .csv data file from the Lick/Hamilton archive

    Args:
        data_file (str): path to .csv file
        SYSTEM_HD (int): HD number of the system

    Returns:
        pd.DataFrame: Pandas dataframe with the standard column names 
        containing Lick/Hamilton data
    """

    raw = pd.read_csv('Lick_Hamilton_F14/' + data_file)
    new = pd.DataFrame()

    new = pd.concat([new, raw[['JD', 'RVel', 'e_RVel']]])

    new['HD'] = SYSTEM_HD
    new['Bibcode'] = '2014yCat..22100005F'
    #JD
    #RVel
    #e_RVel
    new['Sind'] = -1
    new['e_Sind'] = -1
    new['Hind'] = -1
    new['e_Hind'] = -1
    new['EWHa'] = -1
    new['e_EWHa'] = -1
    new['RHKp'] = -1
    new['e_RHKp'] = -1
    new['BIS'] = 0
    new['FWHM'] = -1
    new['Tel'] = 'Hamilton'

    new = new[STD_COL_NAMES]

    return new

def get_Laliotis(data_file, SYSTEM_HD):
    """Read in a .csv data file from the Laliotis archive

    Args:
        data_file (str): path to .csv file
        SYSTEM_HD (int): HD number of the system

    Returns:
        pd.DataFrame: Pandas dataframe with the standard column names 
        containing Laliotis data
    """
        
    raw = pd.read_csv('Laliotis_2023/' + data_file)
    new = pd.DataFrame()

    new = pd.concat([new, raw[['BJD', 'RVel', 'r_RVel', 'e_RVel', 'SHK', 'e_SHK', 'EWHa', 'e_EWHa']]])
    new = new.rename(columns={'BJD':'JD', 'r_RVel':'Tel', 'SHK':'Sind', 'e_SHK':'e_Sind'})

    new['HD'] = SYSTEM_HD
    new['Bibcode'] = '2023yCat..51650176L'
    #JD
    #RVel
    #e_RVel
    new['Sind'] = new['Sind'].fillna(-1)
    new['e_Sind'] = new['e_Sind'].fillna(-1)
    new['Hind'] = -1
    new['e_Hind'] = -1
    new['EWHa'] = new['EWHa'].fillna(-1)
    new['e_EWHa'] = new['e_EWHa'].fillna(-1)
    new['RHKp'] = -1
    new['e_RHKp'] = -1
    new['BIS'] = 0
    new['FWHM'] = -1

    new = new[STD_COL_NAMES]

    return new
```
